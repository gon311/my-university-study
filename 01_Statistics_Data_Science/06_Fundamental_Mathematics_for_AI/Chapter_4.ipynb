# SciPy 이용한 확률분포 계산

```python
from scipy import stats

# 항수 지향적인 방식 : 필요한 값을 바로 넣어 호출
x = stats.binom.rvs(n=5, p=0.2, size=10)
print(x)

# 보다 객체 지향적인 방식 : 분포 객체를 먼저 만들고 사용
dist_binom = stats.binom(n=5, p=0.2)
x = dist_binom.rvs(size=10)
print(x)
```

주요 확률분포 객체 생성 함수(클래스 역할)

| 분포 | 객체 생성 함수(파라미터)|
|-----|-------------------|
| 베르누이분포 |`stats.bernoulli(p)|
| 이항분포 | `stats.binom(n, p) |
| 임의의 이산형 분포 | stats.rv_discrete(values=(xk, pk))  |
| 정규분포 | stats.norm(loc, scale)  |

* 정규분포 loc = 평균, scale= 표준편차


확률분포 객체 주요 메서드
+ rvs(size) : 확률변수의 실현값을 size 만큼 생성
+ pmf(k) : 이산형 확률변수의 확률질량함수 $p_x(k)$ 계산 ($k$는 정수)
+ pdf(x) : 연속형 확률변수의 확률밀도함수 $f_X(x)$ 계산
+ cdf(x) : 확률변수의 $x$에 대한 누적확률$P(X ≤ x)$ 계산
+ ppf(q) : 누적확률 $q$가 주어졌을 때의 분위수($x$값) 계산

  
요약 통계량 메서드
+ mean() : 확률변수의 기댓값 계산 $=E[X]$
+ var() : 확률변수의 분산 계산 $=Var[X]$
+ std() : 확률변수의 표준편차 계산 $=s$



```python
from scipy import stats

# B(5, 0.2) 객체 만들기
B = stats.binom(5, 0.2)
B
# <scipy.stats._distn_infrastructure.rv_discrete_frozen at 0x7aee8581c1a0>
```

```python
# 이산형 확률변수의 확률질량함수
print(B.pmf(0))
print(B.pmf(1))
print(B.pmf(2))
print(B.pmf(3))
print(B.pmf(4))
print(B.pmf(5))
```

```python
# 확률변수의  𝑥 에 대한 누적확률
print(B.cdf(0))
print(B.cdf(1))
print(B.cdf(2))
print(B.cdf(3))
print(B.cdf(4))
print(B.cdf(5))
```

```python
# 역함수 계산
print(B.ppf(q=0.99328))
```

## X의 확률분포를 확률분포표로 구현

```python
# X의 확률분포를 확률분포표로 구현

from scipy import stats

# 확률분포표로 B(5, 0.2) 구현
xk = (0, 1, 2, 3, 4, 5)
pk = (0.32768, 0.4096, 0.2048, 0.0512, 0.0064, 0.00032)

B_custom = stats.rv_discrete(values=(xk, pk))
B_custom
# <scipy.stats._distn_infrastructure.rv_sample at 0x7aee9e042710>
```

```python
print(B_custom.pmf(0))
print(B_custom.pmf(1))
print(B_custom.pmf(2))
print(B_custom.pmf(3))
print(B_custom.pmf(4))
print(B_custom.pmf(5))
```

## X의 기댓값과 분산

```python
# 기댓값 메서드로 계산
print(B.mean())
# 기댓값 공식으로 직접 계산
E_X = 0*B.pmf(0)+1*B.pmf(1)+2*B.pmf(2)+3*B.pmf(3)+4*B.pmf(4)+5*B.pmf(5)
print(E_X)
```

```python
# 분산을 메서드로 계산
print(B.var())
# 분산을 공식으로 직접 계산
E_X2 = 1**2*B.pmf(1)+2**2*B.pmf(2)+3**2*B.pmf(3)+4**2*B.pmf(4)+5**2*B.pmf(5)
Var = E_X2 - E_X**2
print(Var)
```

## X의 기댓값과 분산 & (표본)평균과 (표본)분산

```python
from scipy import stats

N = stats.norm(loc=0, scale=1/3)


# 이론적 기댓값과 분산 (N 사용)
theo_mean = N.mean()
theo_var = N.var()
print(f"이론적 평균:            {theo_mean} & 이론적 분산 : {theo_var:.4f}")


#  표본(simple)을 통한 (표본)평균과 (표본)분산
N_sample = N.rvs(100)
sample_mean = N_sample.mean()
sample_var = N_sample.var(ddof=1)
print(f"표본 100개의 평균 : {sample_mean:.4f} & 분산 : {sample_var:.4f}")
```

```python
N_sample = N.rvs(1000)
sample_mean = N_sample.mean()
sample_var = N_sample.var(ddof=1)
print(f"표본 1000개의 평균 : {sample_mean:.4f} & 분산 : {sample_var:.4f}")
```

```python
N_sample = N.rvs(10000)
sample_mean = N_sample.mean()
sample_var = N_sample.var(ddof=1)
print(f"표본 10000개의 평균 : {sample_mean:.4f} & 분산 : {sample_var:.4f}")
```
